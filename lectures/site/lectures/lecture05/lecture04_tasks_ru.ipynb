## Задание 3. Оценка по баллам

Дан балл от 0 до 100. Выведите оценку:
- 90-100: "A"
- 75-89:  "B"
- 60-74:  "C"
- 0-59:   "F"
- иначе:  "некорректный балл"

Проверьте: 95, 75, 60, 40, 150.

````python


scores = [95,75,60,40,150]


def grade(score):
    if 90 <= score <= 100:
        return "A"
    elif 75 <= score < 90:
        return 'B'
    elif 60 <= score < 75:
        return 'C'
    elif 0 <= score < 60:
        return 'F'
    else:
        return 'Некоректный балл'
    
for i in scores:
    print(f'{i}={grade(i)}')






````
